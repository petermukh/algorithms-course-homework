# Задача 2. Iterative

Реализовать итеративные версии mergesort и quicksort.

## Идея

В итеративном mergesort сначала считаем отсортированными участки длины `1`. Сливаем соседние участки, затем удваиваем их длину: `1 → 2 → 4 → ...`, пока не обработаем весь массив.

В итеративном quicksort вместо рекурсивных вызовов храним границы ещё не обработанных частей в обычном списке-стеке. Извлекаем очередную часть, ставим опорный элемент на место и добавляем в стек левую и правую части. В обоих случаях исходный список не меняется.

```python
def iterative_mergesort(nums):
    source = nums[:]
    n = len(source)
    buffer = [None] * n
    width = 1

    while width < n:
        for left in range(0, n, 2 * width):
            middle = min(left + width, n)
            right = min(left + 2 * width, n)
            i, j, out = left, middle, left

            while i < middle and j < right:
                if source[i] <= source[j]:
                    buffer[out] = source[i]
                    i += 1
                else:
                    buffer[out] = source[j]
                    j += 1
                out += 1

            while i < middle:
                buffer[out] = source[i]
                i += 1
                out += 1
            while j < right:
                buffer[out] = source[j]
                j += 1
                out += 1

        source, buffer = buffer, source
        width *= 2

    return source


def iterative_quicksort(nums):
    result = nums[:]

    def partition(left, right):
        pivot = result[right]
        border = left
        for i in range(left, right):
            if result[i] <= pivot:
                result[border], result[i] = result[i], result[border]
                border += 1
        result[border], result[right] = result[right], result[border]
        return border

    stack = [(0, len(result) - 1)]
    while stack:
        left, right = stack.pop()
        if left >= right:
            continue
        pivot_index = partition(left, right)
        if left < pivot_index - 1:
            stack.append((left, pivot_index - 1))
        if pivot_index + 1 < right:
            stack.append((pivot_index + 1, right))

    return result
```

## Пошаговый пример

Mergesort для `[4, 1, 3, 2]`:

| Длина участков перед слиянием | Состояние после прохода |
|---:|:---|
| 1 | `[1, 4, 2, 3]` |
| 2 | `[1, 2, 3, 4]` |

Quicksort для того же массива сначала берёт границы `(0, 3)`. Опорное значение `2` попадает на позицию `1`, массив становится `[1, 2, 3, 4]`. В стеке остаётся правая часть `(2, 3)`; после её обработки стек пуст, сортировка закончена.

## Сложность

Итеративный mergesort делает `O(log n)` проходов по `n` элементам: время `O(n log n)`, память `O(n)` на рабочий буфер и копию. Итеративный quicksort в среднем занимает `O(n log n)` времени, в худшем случае `O(n²)` из-за неудачных опорных элементов. Стек диапазонов может занимать до `O(n)` памяти в худшем случае, плюс `O(n)` на копию входа; рекурсивного стека нет.

## Тесты

Проверяем разные размеры, повторы, отрицательные значения и отсутствие изменения входного списка. Большой отсортированный массив показывает, что итеративный quicksort не зависит от лимита рекурсии.

```python
cases = [
    [], [7], [2, 1], [4, 1, 3, 2], [5, 5, 5, 5],
    [-3, 8, 0, -1, 8], list(range(30)), list(range(30, 0, -1))
]
for case in cases:
    expected = sorted(case)
    before = case[:]
    assert iterative_mergesort(case) == expected
    assert iterative_quicksort(case) == expected
    assert case == before

long_ordered = list(range(1200))
assert iterative_mergesort(long_ordered) == long_ordered
assert iterative_quicksort(long_ordered) == long_ordered
print("Все тесты пройдены")
```
