# Задача 3. K-th

Найти `k`-й по величине элемент массива. Нельзя просто отсортировать массив и нельзя использовать кучу. Подсказка: quickselect.

Примеры:

- `nums = [3, 2, 1, 5, 6, 4]`, `k = 2` → `5`
- `nums = [3, 2, 3, 1, 2, 4, 5, 5, 6]`, `k = 4` → `4`

## Идея

Если бы массив был отсортирован по возрастанию, нужный элемент стоял бы на индексе `len(nums) - k`. Но сортировать всё не нужно. Quickselect выбирает опорное значение и делит текущую часть массива на три группы: меньше, равно и больше опорного. Затем продолжает работу только с той группой, где находится нужный индекс. Группа равных значений помогает быстро обработать повторы.

Функция работает с копией списка, поэтому вход не меняется. При неправильном `k` она выдаёт `ValueError`. Необязательный параметр `show_steps=True` печатает состояния для пошагового примера.

```python
def kth_largest(nums, k, show_steps=False):
    if not 1 <= k <= len(nums):
        raise ValueError("k должно быть от 1 до длины массива")

    values = nums[:]
    target = len(values) - k
    left, right = 0, len(values) - 1

    while left <= right:
        pivot = values[(left + right) // 2]
        less = left
        current = left
        greater = right

        while current <= greater:
            if values[current] < pivot:
                values[less], values[current] = values[current], values[less]
                less += 1
                current += 1
            elif values[current] > pivot:
                values[current], values[greater] = values[greater], values[current]
                greater -= 1
            else:
                current += 1

        if show_steps:
            print(f"pivot={pivot}, меньше: {left}..{less - 1}, "
                  f"равно: {less}..{greater}, больше: {greater + 1}..{right}")
            print("состояние:", values)

        if target < less:
            right = less - 1
        elif target > greater:
            left = greater + 1
        else:
            return values[target]
```

## Пошаговый пример

Для `[3, 2, 1, 5, 6, 4]` и `k = 2` целевой индекс — `6 - 2 = 4`. На каждом шаге смотрим, в какую из трёх групп попал индекс `4`, и отбрасываем остальные. Вывод следующей ячейки показывает реальные состояния списка после разделения; порядок внутри групп не обязан быть отсортированным.

```python
print("Ответ:", kth_largest([3, 2, 1, 5, 6, 4], 2, show_steps=True))
```

## Сложность

За одно разделение просматриваем текущую часть массива один раз. В среднем остаётся существенно меньшая часть, поэтому ожидаемое время — `O(n)`; при очень неудачных опорных значениях худший случай — `O(n²)`. Копия входа занимает `O(n)` памяти, сам алгоритм дополнительно хранит лишь несколько индексов — `O(1)`. Полная дополнительная память этой реализации — `O(n)`.

## Тесты

Проверяем оба примера, крайние значения `k`, повторы, отрицательные числа, одинаковые элементы, неизменность входа и ошибку для недопустимого `k`. `sorted` используется только в тесте для проверки ответа, не внутри решения.

```python
assert kth_largest([3, 2, 1, 5, 6, 4], 2) == 5
assert kth_largest([3, 2, 3, 1, 2, 4, 5, 5, 6], 4) == 4
assert kth_largest([7], 1) == 7
assert kth_largest([5, 5, 5, 5], 3) == 5
assert kth_largest([-3, -1, -2], 1) == -1
assert kth_largest([-3, -1, -2], 3) == -3

original = [8, 1, 8, 2, 3]
for k in range(1, len(original) + 1):
    assert kth_largest(original, k) == sorted(original, reverse=True)[k - 1]
assert original == [8, 1, 8, 2, 3]

for bad_k in (0, 4):
    try:
        kth_largest([1, 2, 3], bad_k)
    except ValueError:
        pass
    else:
        raise AssertionError("Для неправильного k ожидалась ошибка")
try:
    kth_largest([], 1)
except ValueError:
    pass
else:
    raise AssertionError("Для пустого массива ожидалась ошибка")

print("Все тесты пройдены")
```
