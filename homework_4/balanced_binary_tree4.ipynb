## Задача 3. Balanced Binary Tree

Реализуйте алгоритм, который определяет, является ли бинарное дерево сбалансированным по высоте.

Дерево считается сбалансированным по высоте, если для каждого узла высоты левого и правого поддеревьев отличаются не более чем на 1.

Результат — True или False.

### Идея решения

Сначала узнаём высоты левого и правого поддеревьев, затем проверяем их разницу. Это обход снизу вверх. В этой реализации пустое поддерево имеет высоту `0`, лист — высоту `1`. В лекции высота считается как глубина самого глубокого узла, поэтому числа отличаются на единицу, но проверка разницы высот работает одинаково.

Вспомогательная функция возвращает высоту поддерева, если оно сбалансировано, и `-1`, если найдено нарушение. Получив `-1`, проверку можно прекратить. Высоту каждого узла считаем только один раз.

### Реализация

````python
class TreeNode:
    def __init__(self, value, left=None, right=None):
        self.value = value
        self.left = left
        self.right = right


def is_balanced(root):
    def height_or_error(node):
        if node is None:
            return 0

        left_height = height_or_error(node.left)
        if left_height == -1:
            return -1

        right_height = height_or_error(node.right)
        if right_height == -1:
            return -1

        if abs(left_height - right_height) > 1:
            return -1

        return 1 + max(left_height, right_height)

    return height_or_error(root) != -1
````

### Оценка сложности

Время — `O(n)`: высота каждого посещённого узла вычисляется один раз. Дополнительная память — `O(h)` на стек рекурсии; в худшем случае `h = n`, а для сбалансированного дерева `h = O(log n)`.

### Визуализация работы

У корня `10` высоты обоих поддеревьев равны `3`, но этого недостаточно: проверять нужно **каждый** узел.

```text
        10
       /  \
      5    15
     /      \
    3        17
   /          \
  2            20
```

| Узел | Высота слева | Высота справа | Вывод |
|---:|---:|---:|:---|
| `2` | 0 | 0 | высота 1 |
| `3` | 1 | 0 | высота 2 |
| `5` | 2 | 0 | разница 2: нарушение |

После обнаружения нарушения функция передаёт `-1` вверх и возвращает `False`.

### Тесты

Проверяем пустое дерево, одиночный узел, сбалансированное дерево, цепочку, нарушение ниже корня и дерево, которое не является BST, но сбалансировано по высоте.

````python
balanced_root = TreeNode(5,
    TreeNode(3, TreeNode(2), TreeNode(4)),
    TreeNode(7, TreeNode(6), TreeNode(8)))

assert is_balanced(None) is True
assert is_balanced(TreeNode(1)) is True
assert is_balanced(balanced_root) is True
assert is_balanced(TreeNode(1, TreeNode(2), None)) is True
assert is_balanced(TreeNode(1, TreeNode(2, TreeNode(3)))) is False
hidden_imbalance = TreeNode(10,
    TreeNode(5, TreeNode(3, TreeNode(2))),
    TreeNode(15, right=TreeNode(17, right=TreeNode(20))))
assert is_balanced(hidden_imbalance) is False
assert is_balanced(TreeNode(5, TreeNode(9), TreeNode(1))) is True

print('Задача 3: все тесты пройдены')
````
