## Задача 2. Validate BST

Реализуйте алгоритм, который проверяет, является ли произвольное бинарное дерево корректным деревом поиска (BST).

Результат — True или False.

### Идея решения

Обходим дерево в порядке inorder: сначала левое поддерево, затем узел, затем правое. У корректного BST значения при таком обходе идут строго по возрастанию. Держим предыдущее посещённое значение и сравниваем с текущим. Если текущий узел не больше предыдущего, возвращаем `False`.

Проверять только ближайших детей недостаточно: нарушение может быть глубже. Inorder-обход обнаружит его, потому что порядок значений перестанет быть строго возрастающим. Стек хранит узлы, к которым нужно вернуться после обхода левого поддерева.

### Реализация

````python
class TreeNode:
    def __init__(self, value, left=None, right=None):
        self.value = value
        self.left = left
        self.right = right


def is_valid_bst(root):
    stack = []
    current = root
    previous = None

    while current is not None or stack:
        while current is not None:
            stack.append(current)
            current = current.left

        current = stack.pop()
        if previous is not None and current.value <= previous:
            return False
        previous = current.value
        current = current.right

    return True
````

### Оценка сложности

В худшем случае проверка занимает `O(n)` времени: каждый узел кладём в стек и извлекаем из него по одному разу. Дополнительная память — `O(h)` на стек обхода, где `h` — высота дерева. Для сбалансированного дерева это `O(log n)`, для цепочки — `O(n)`.

### Визуализация работы

В этом дереве `6` находится в левом поддереве корня `5`, хотя `6 > 5`:

```text
      5
     / \
    3   7
     \
      6
```

| Шаг inorder | Текущий узел | Предыдущее значение | Результат |
|---:|---:|---:|:---|
| 1 | `3` | нет | запоминаем `3` |
| 2 | `6` | `3` | `6 > 3`, запоминаем `6` |
| 3 | `5` | `6` | `5 ≤ 6`: порядок нарушен |

Значит, функция возвращает `False`.

### Тесты

Проверяем пустое дерево, корректное BST, нарушение у прямого ребёнка, нарушение из-за дальнего предка, дубликат, отрицательные значения и цепочку из четырёх узлов.

````python
valid_root = TreeNode(5,
    TreeNode(3, TreeNode(2), TreeNode(4)),
    TreeNode(7, TreeNode(6), TreeNode(8)))
chain_root = TreeNode(1, right=TreeNode(2, right=TreeNode(3, right=TreeNode(4))))

assert is_valid_bst(None) is True
assert is_valid_bst(TreeNode(5)) is True
assert is_valid_bst(valid_root) is True
assert is_valid_bst(TreeNode(5, TreeNode(8), TreeNode(9))) is False
deep_violation = TreeNode(5, TreeNode(3, right=TreeNode(6)), TreeNode(7))
assert is_valid_bst(deep_violation) is False
assert is_valid_bst(TreeNode(5, TreeNode(3), TreeNode(5))) is False
assert is_valid_bst(TreeNode(0, TreeNode(-10), TreeNode(10))) is True
assert is_valid_bst(chain_root) is True

print('Задача 2: все тесты пройдены')
````
