# Домашнее задание №4 — задача 1. Tree Traversal

## Задача 1. Tree Traversal

1. Реализуйте бинарное дерево поиска (BST) и операцию вставки нового элемента с сохранением свойств BST.
2. Реализуйте шесть обходов дерева: preorder, inorder, postorder, reverse preorder, reverse inorder, reverse postorder.

Каждый обход должен возвращать список значений узлов в порядке посещения.

Источник с обозначениями обходов: [Tree traversal](https://en.wikipedia.org/wiki/Tree_traversal).

### Идея решения

При вставке сравниваем новое значение с текущим узлом: меньше — идём налево, больше — направо. Когда соответствующая ссылка пуста, создаём узел. Равное значение не вставляем. Так сохраняется строгий порядок BST.

Для обхода достаточно трёх действий: посетить узел (`N`), обойти левое поддерево (`L`) и правое (`R`). Шесть обходов отличаются только порядком этих действий:

| Обход | Порядок |
|:---|:---:|
| preorder | `NLR` |
| inorder | `LNR` |
| postorder | `LRN` |
| reverse preorder | `NRL` |
| reverse inorder | `RNL` |
| reverse postorder | `RLN` |

Рекурсивная функция `_traverse` выполняет эти три действия в указанном порядке. Для пустого поддерева она ничего не добавляет.

### Реализация

````python
class TreeNode:
    def __init__(self, value, left=None, right=None):
        self.value = value
        self.left = left
        self.right = right


class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, value):
        if self.root is None:
            self.root = TreeNode(value)
            return

        current = self.root
        while True:
            if value < current.value:
                if current.left is None:
                    current.left = TreeNode(value)
                    return
                current = current.left
            elif value > current.value:
                if current.right is None:
                    current.right = TreeNode(value)
                    return
                current = current.right
            else:
                return  # повторное значение не добавляем

    def _traverse(self, order):
        result = []

        def visit(node):
            if node is None:
                return

            for step in order:
                if step == 'N':
                    result.append(node.value)
                elif step == 'L':
                    visit(node.left)
                else:  # step == 'R'
                    visit(node.right)

        visit(self.root)
        return result

    def preorder(self):
        return self._traverse('NLR')

    def inorder(self):
        return self._traverse('LNR')

    def postorder(self):
        return self._traverse('LRN')

    def reverse_preorder(self):
        return self._traverse('NRL')

    def reverse_inorder(self):
        return self._traverse('RNL')

    def reverse_postorder(self):
        return self._traverse('RLN')
````

### Оценка сложности

Пусть `n` — количество узлов, а `h` — высота дерева. Одна вставка занимает `O(h)` времени и `O(1)` дополнительной памяти: цикл идёт по одному пути от корня к месту вставки. Если дерево сбалансировано, `h = O(log n)`; если превратилось в цепочку, `h = O(n)`.

Любой обход занимает `O(n)` времени: каждый узел посещается один раз. Память — `O(n + h)`: `O(n)` на возвращаемый список и `O(h)` на стек рекурсивных вызовов. Если не считать обязательный список результата, дополнительная память равна `O(h)`.

### Визуализация работы

Вставляем `5, 3, 7, 2, 4, 6, 8`. Для `4` путь вставки: `4 < 5`, идём влево; `4 > 3`, идём вправо; правая ссылка пуста — ставим узел `4`. Аналогично получаем дерево:

```text
        5
      /   \
     3     7
    / \   / \
   2   4 6   8
```

| Обход | Порядок посещения |
|:---|:---|
| preorder (`NLR`) | `5, 3, 2, 4, 7, 6, 8` |
| inorder (`LNR`) | `2, 3, 4, 5, 6, 7, 8` |
| postorder (`LRN`) | `2, 4, 3, 6, 8, 7, 5` |
| reverse preorder (`NRL`) | `5, 7, 8, 6, 3, 4, 2` |
| reverse inorder (`RNL`) | `8, 7, 6, 5, 4, 3, 2` |
| reverse postorder (`RLN`) | `8, 6, 7, 4, 2, 3, 5` |

### Тесты

Проверяем пустое дерево, один узел, все шесть обходов, сохранение сортировки после вставок и повторное значение.

````python
empty_tree = BinarySearchTree()
for traversal in (empty_tree.preorder, empty_tree.inorder, empty_tree.postorder,
                  empty_tree.reverse_preorder, empty_tree.reverse_inorder,
                  empty_tree.reverse_postorder):
    assert traversal() == []

single = BinarySearchTree()
single.insert(5)
assert single.inorder() == [5]
assert single.reverse_postorder() == [5]
single.insert(5)
assert single.inorder() == [5]  # дубликат пропущен

tree = BinarySearchTree()
for value in (5, 3, 7, 2, 4, 6, 8):
    tree.insert(value)
assert tree.preorder() == [5, 3, 2, 4, 7, 6, 8]
assert tree.inorder() == [2, 3, 4, 5, 6, 7, 8]
assert tree.postorder() == [2, 4, 3, 6, 8, 7, 5]
assert tree.reverse_preorder() == [5, 7, 8, 6, 3, 4, 2]
assert tree.reverse_inorder() == [8, 7, 6, 5, 4, 3, 2]
assert tree.reverse_postorder() == [8, 6, 7, 4, 2, 3, 5]

chain = BinarySearchTree()
for value in (1, 2, 3, 4):
    chain.insert(value)
assert chain.inorder() == [1, 2, 3, 4]
assert chain.postorder() == [4, 3, 2, 1]

print('Задача 1: все тесты пройдены')
````
