## **Задача 2. Сумма**

На вход подается массив целых положительных чисел, разделенных пробелом.

Нужно найти максимальную сумму элементов массива, которая делится на 2.

Требования:

* Напишите тесты, которые проверяют функцию на разных случаях.
* Реализовать тесты можно любым способом (unittest, pytest)

Формат сдачи: в своем репозитории /homework_1/sum/

**Пример 1:**

Вход: 5 7 13 2 14

Выход: 36

Пояснение: Берем 2, 7, 13, 14

**Пример 2:**

Вход: 3

Выход: 0

Пояснение: 3 не делится на 2, поэтому в ответе 0.

### Идея решения

Поскольку все числа положительные, выгодно сначала взять все элементы массива.

- Если общая сумма чётная, это и есть максимальный ответ.
- Если общая сумма нечётная, в выбранных элементах находится нечётное количество нечётных чисел. Чтобы сумма стала чётной и уменьшилась как можно меньше, нужно удалить самое маленькое нечётное число.

Так мы получаем наибольшую возможную чётную сумму.

### Реализация

```python
def max_even_sum(numbers: list[int]) -> int:
    total_sum = 0
    smallest_odd = None

    for number in numbers:
        total_sum += number

        if number % 2 != 0:
            if smallest_odd is None or number < smallest_odd:
                smallest_odd = number

    if total_sum % 2 == 0:
        return total_sum

    return total_sum - smallest_odd
```

### Почему решение корректно

Если сумма всех элементов чётная, никакое подмножество не может иметь большую сумму, поскольку все элементы положительные.

Если сумма нечётная, для изменения её чётности необходимо исключить нечётную сумму. Наименьшая возможная положительная нечётная сумма, которую можно исключить, — минимальный нечётный элемент массива. Удаление любого другого набора с нечётной суммой уменьшит результат не меньше.

### Оценка сложности

- **Время:** `O(n)`, потому что массив просматривается один раз.
- **Дополнительная память:** `O(1)`, потому что кроме входного массива используются только несколько переменных.

### Визуализация работы алгоритма

Рассмотрим массив `[5, 7, 13, 2, 14]`:

| Прочитан элемент | Текущая сумма | Минимальный нечётный |
|---:|---:|---:|
| 5 | 5 | 5 |
| 7 | 12 | 5 |
| 13 | 25 | 5 |
| 2 | 27 | 5 |
| 14 | 41 | 5 |

Общая сумма `41` нечётная. Удаляем минимальное нечётное число `5`: `41 - 5 = 36`.

### Тесты

Проверим примеры из условия, пустой массив, одно чётное число, уже чётную общую сумму, набор нечётных чисел и несколько нетривиальных случаев.

```python
assert max_even_sum([5, 7, 13, 2, 14]) == 36  # пример из условия
assert max_even_sum([3]) == 0                  # пример из условия
assert max_even_sum([]) == 0                   # пустой массив
assert max_even_sum([2]) == 2                  # один чётный элемент
assert max_even_sum([2, 4, 6]) == 12           # все элементы чётные
assert max_even_sum([1, 3, 5]) == 8            # удаляем минимальное нечётное
assert max_even_sum([8, 1, 2, 3]) == 14        # общая сумма уже чётная
assert max_even_sum([9, 2, 7, 4, 1]) == 22     # удаляем 1 из суммы 23

print("Все тесты пройдены")
```
