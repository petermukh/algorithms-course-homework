## **Задача 3. Простые числа**

На вход подается целое число N

Нужно найти количество простых чисел < N.

Требования:

* Напишите тесты, которые проверяют функцию на разных случаях.
* Реализовать тесты можно любым способом (unittest, pytest)

Формат сдачи: в своем репозитории /homework_1/prime/

**Пример 1:**

Вход: 10

Выход: 4

Пояснение: 2, 3, 5, 7

**Пример 2:**

Вход: 1

Выход: 0

### Идея решения

Переберём все числа от `2` до `N - 1`. Для каждого числа проверим, есть ли у него делитель, кроме 1 и самого числа.

Делители будем проверять начиная с 2. Если число делится на очередной делитель без остатка, оно не является простым. Если подходящего делителя не нашлось, увеличиваем счётчик простых чисел.

Проверять делители достаточно, пока квадрат делителя не превышает проверяемое число. Если число составное, хотя бы один из двух его множителей обязательно будет не больше квадратного корня из числа.

### Реализация

```python
def is_prime(number: int) -> bool:
    if number < 2:
        return False

    divisor = 2

    while divisor * divisor <= number:
        if number % divisor == 0:
            return False

        divisor += 1

    return True


def count_primes(n: int) -> int:
    primes_count = 0

    for number in range(2, n):
        if is_prime(number):
            primes_count += 1

    return primes_count
```

### Оценка сложности

- **Время:** `O(n √n)`. Мы проверяем до `n` чисел, а для каждого из них перебираем не более `√n` возможных делителей.
- **Память:** `O(1)`, потому что используется постоянное количество переменных.

### Визуализация работы алгоритма

Рассмотрим `N = 10`. Проверяем числа от 2 до 9:

| Число | Проверка | Простое? | Количество простых |
|---:|:---|:---:|---:|
| 2 | Делителей проверять не нужно | Да | 1 |
| 3 | Делителей проверять не нужно | Да | 2 |
| 4 | Делится на 2 | Нет | 2 |
| 5 | Не делится на 2 | Да | 3 |
| 6 | Делится на 2 | Нет | 3 |
| 7 | Не делится на 2 | Да | 4 |
| 8 | Делится на 2 | Нет | 4 |
| 9 | Делится на 3 | Нет | 4 |

Простые числа меньше 10: `2, 3, 5, 7`. Их количество равно 4.

### Тесты

Проверим примеры из условия, значения около первого простого числа, небольшие и более крупные границы.

```python
assert count_primes(10) == 4     # 2, 3, 5, 7
assert count_primes(1) == 0      # пример из условия
assert count_primes(0) == 0      # чисел меньше 0 нет
assert count_primes(2) == 0      # простые строго меньше N
assert count_primes(3) == 1      # только 2
assert count_primes(4) == 2      # 2 и 3
assert count_primes(20) == 8     # несколько простых чисел
assert count_primes(100) == 25   # более крупный случай

print("Все тесты пройдены")
```
