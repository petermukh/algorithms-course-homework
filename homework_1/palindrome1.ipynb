## **Задача 1. Палиндром**

Напишите программу, которая проверяет, является ли целое положительное число палиндромом.

Сделайте это без использования строк.

Требования:

* Напишите тесты, которые проверяют функцию на разных случаях.
* Реализовать тесты можно любым способом (unittest, pytest)

Формат сдачи: в своем репозитории /homework_1/palindrome/

**Пример 1:**

Вход: 121

Выход: True

**Пример 2:**

Вход: 31

Выход: False

### Идея решения

Сохраним исходное число. Затем будем по одной извлекать его последние цифры с помощью остатка от деления на 10. Из извлечённых цифр соберём число в обратном порядке.

На каждом шаге:

1. `number % 10` получает последнюю цифру;
2. `reversed_number * 10` освобождает место для новой цифры;
3. `number //= 10` удаляет последнюю цифру исходного числа.

Если перевёрнутое число совпадает с исходным, число является палиндромом. Строки в алгоритме не используются.

### Реализация

```python
def is_palindrome(number: int) -> bool:
    if number < 0:
        return False

    original_number = number
    reversed_number = 0

    while number != 0:
        reversed_number = 10 * reversed_number + number % 10
        number //= 10

    return reversed_number == original_number
```

### Оценка сложности

- **Время:** `O(log n)`, потому что количество итераций равно количеству цифр числа.
- **Память:** `O(1)`, потому что используется постоянное количество переменных независимо от величины числа.

### Визуализация работы алгоритма

Рассмотрим число `12321`:

| Шаг | `number` до удаления цифры | Последняя цифра | `reversed_number` |
|---:|---:|---:|---:|
| 1 | 12321 | 1 | 1 |
| 2 | 1232 | 2 | 12 |
| 3 | 123 | 3 | 123 |
| 4 | 12 | 2 | 1232 |
| 5 | 1 | 1 | 12321 |

Получили `12321`, оно равно исходному числу, поэтому функция возвращает `True`.

### Тесты

Проверим примеры из условия, однозначное число, числа с чётным и нечётным количеством цифр, число с нулём внутри и непалиндромы.

```python
assert is_palindrome(121) is True       # пример из условия
assert is_palindrome(31) is False        # пример из условия
assert is_palindrome(7) is True          # однозначное число
assert is_palindrome(1221) is True       # чётное количество цифр
assert is_palindrome(12321) is True      # нечётное количество цифр
assert is_palindrome(1001) is True       # нули внутри числа
assert is_palindrome(10) is False        # число заканчивается нулём
assert is_palindrome(12345) is False     # нетривиальный непалиндром

print("Все тесты пройдены")
```
