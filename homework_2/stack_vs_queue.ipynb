## **Stack vs queue**

Реализовать стек и очередь на основе связных списков. Без использования сторонних библиотек.

### Идея решения

Связный список состоит из узлов. Каждый узел хранит значение и ссылку на следующий узел.

- В **стеке** мы добавляем и удаляем элементы только с начала списка. Поэтому `push` и `pop` работают по принципу LIFO: последним добавили — первым получили.
- В **очереди** храним ссылки и на начало, и на конец списка. Новый элемент добавляется в конец, а удаляется из начала. Поэтому `enqueue` и `dequeue` работают по принципу FIFO: первым добавили — первым получили.

Для пустой структуры методы `pop`, `peek` и `dequeue` выбрасывают `IndexError`: получить элемент из пустого стека или очереди невозможно.

### Реализация

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None


class LinkedStack:
    def __init__(self):
        self.head = None
        self.size = 0

    def push(self, value):
        new_node = Node(value)
        new_node.next = self.head
        self.head = new_node
        self.size += 1

    def pop(self):
        if self.head is None:
            raise IndexError("Stack is empty")

        value = self.head.value
        self.head = self.head.next
        self.size -= 1
        return value

    def peek(self):
        if self.head is None:
            raise IndexError("Stack is empty")

        return self.head.value

    def is_empty(self):
        return self.head is None

    def __len__(self):
        return self.size


class LinkedQueue:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def enqueue(self, value):
        new_node = Node(value)

        if self.tail is None:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            self.tail = new_node

        self.size += 1

    def dequeue(self):
        if self.head is None:
            raise IndexError("Queue is empty")

        value = self.head.value
        self.head = self.head.next
        self.size -= 1

        if self.head is None:
            self.tail = None

        return value

    def peek(self):
        if self.head is None:
            raise IndexError("Queue is empty")

        return self.head.value

    def is_empty(self):
        return self.head is None

    def __len__(self):
        return self.size
```

### Оценка сложности

| Операция | Время | Почему |
|:---|:---:|:---|
| `stack.push` | `O(1)` | Добавляем узел в начало списка. |
| `stack.pop` | `O(1)` | Удаляем первый узел. |
| `queue.enqueue` | `O(1)` | Есть ссылка на последний узел. |
| `queue.dequeue` | `O(1)` | Удаляем первый узел. |

Память — `O(n)`: для `n` добавленных элементов хранится `n` узлов.

### Визуализация работы

Добавим одинаковые значения `10`, `20`, `30` в стек и очередь.

| Действие | Стек (верх слева) | Очередь (начало слева) |
|:---|:---|:---|
| Добавить 10 | `10` | `10` |
| Добавить 20 | `20 → 10` | `10 → 20` |
| Добавить 30 | `30 → 20 → 10` | `10 → 20 → 30` |
| Удалить элемент | `pop()` вернёт `30` | `dequeue()` вернёт `10` |
| Остаток | `20 → 10` | `20 → 30` |

Стек возвращает последний добавленный элемент, а очередь — первый добавленный.

### Тесты

Проверим порядок извлечения, работу `peek`, длину, пустые структуры и повторное использование очереди после её очистки.

```python
# Тесты стека
stack = LinkedStack()
assert stack.is_empty() is True
assert len(stack) == 0

stack.push(10)
stack.push(20)
stack.push(30)
assert stack.peek() == 30
assert len(stack) == 3
assert stack.pop() == 30
assert stack.pop() == 20
assert stack.pop() == 10
assert stack.is_empty() is True

try:
    stack.pop()
    assert False
except IndexError:
    pass

# Тесты очереди
queue = LinkedQueue()
assert queue.is_empty() is True

queue.enqueue(10)
queue.enqueue(20)
queue.enqueue(30)
assert queue.peek() == 10
assert len(queue) == 3
assert queue.dequeue() == 10
assert queue.dequeue() == 20
assert queue.dequeue() == 30
assert queue.is_empty() is True

queue.enqueue(42)
assert queue.dequeue() == 42

try:
    queue.dequeue()
    assert False
except IndexError:
    pass

print("Все тесты пройдены")
```
