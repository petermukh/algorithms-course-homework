## **Validate**

Даны 2 последовательности pushed и popped, содержащие уникальные целые числа. popped является перестановкой pushed, то есть, все элементы совпадают, но может отличаться порядок.

Программа должна вернуть True, если эти последовательности могут получиться в результате некоторой последовательности операций push и pop на пустом стеке.

1 <= len(pushed) <= 100_000

**Пример 1:**

pushed: 1 2 3 4 5

popped: 1 3 5 4 2

result: True

Последовательность операций: push(1), pop(1), push(2), push(3), pop(3), push(4), push(5), pop(5), pop(4), pop(2)

**Пример 2:**

pushed: 1 2 3

popped: 3 1 2

result: False

Последовательность операций: push(1), push(2), push(3), pop(3), нужно pop(1), но на вершине стека лежит 2.

**Требования:**

* стороннике библиотеки не используем
* тесты

Формат сдачи: в своем репозитории /homework_2/validate/

### Идея решения

Будем повторять работу настоящего стека. Перебираем `pushed` и добавляем каждое число в обычный список `stack`. Верх стека находится в конце списка.

После каждого добавления проверяем: совпадает ли верх стека со следующим нужным числом из `popped`. Если совпадает, сразу удаляем элемент. Повторяем это, пока удаление возможно.

Если к концу обработки удалось получить всю последовательность `popped`, ответ `True`. Если в какой-то момент нужное число нельзя снять с вершины, в конце останутся необработанные элементы и ответ будет `False`.

### Реализация

```python
def validate_stack_sequences(pushed: list[int], popped: list[int]) -> bool:
    if len(pushed) != len(popped):
        return False

    stack = []
    popped_index = 0

    for number in pushed:
        stack.append(number)

        while (
            stack
            and popped_index < len(popped)
            and stack[-1] == popped[popped_index]
        ):
            stack.pop()
            popped_index += 1

    return popped_index == len(popped)
```

### Оценка сложности

- **Время:** `O(n)`, где `n` — длина последовательности. Каждый элемент один раз добавляется в стек и не более одного раза удаляется из него.
- **Память:** `O(n)`. В худшем случае все элементы `pushed` могут одновременно находиться в стеке.

### Визуализация работы алгоритма

Рассмотрим `pushed = [1, 2, 3, 4, 5]` и `popped = [1, 3, 5, 4, 2]`. Верх стека — справа.

| Действие | Стек | Следующее нужное значение из `popped` |
|:---|:---|:---|
| `push(1)`, затем `pop()` | `[]` | 3 |
| `push(2)` | `[2]` | 3 |
| `push(3)`, затем `pop()` | `[2]` | 5 |
| `push(4)` | `[2, 4]` | 5 |
| `push(5)`, затем `pop()` | `[2, 4]` | 4 |
| `pop()` | `[2]` | 2 |
| `pop()` | `[]` | последовательность закончилась |

Все значения удалось снять в нужном порядке, поэтому функция вернёт `True`.

### Тесты

Проверим оба примера, один элемент, уже обратный порядок, прямой порядок, случай без возможности получить нужный порядок и случай с разной длиной последовательностей.

```python
assert validate_stack_sequences([1, 2, 3, 4, 5], [1, 3, 5, 4, 2]) is True
assert validate_stack_sequences([1, 2, 3], [3, 1, 2]) is False
assert validate_stack_sequences([1], [1]) is True
assert validate_stack_sequences([1, 2, 3], [3, 2, 1]) is True
assert validate_stack_sequences([1, 2, 3], [1, 2, 3]) is True
assert validate_stack_sequences([1, 2, 3, 4], [2, 4, 1, 3]) is False
assert validate_stack_sequences([1, 2], [2]) is False

print("Все тесты пройдены")
```
