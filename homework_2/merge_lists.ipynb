## **Merge lists**

Даны два отсортированных односвязных списка list1 и list2

Необходимо объединить их в один новый отсортированный список.
Новый список должен быть составлен слиянием узлов двух исходных списков.
Вернуть необходимо голову объединенного списка.

Вход: list1 = [1,2,4], list2 = [1,3,4]

Выход: [1,1,2,3,4,4]

**Требования:**

* реализовать два способа решения поставленной задачи: с использованием фиктивного элемента и без него
* написать тесты

Формат сдачи: в своем репозитории /homework_2/merge_lists/

### Идея решения

В обоих вариантах сравниваем первые ещё не добавленные узлы `list1` и `list2`. Меньший узел присоединяем к концу результата и передвигаем указатель в соответствующем списке. Когда один список заканчивается, присоединяем оставшуюся часть второго: она уже отсортирована.

Вариант с фиктивным узлом использует дополнительный начальный узел `dummy`. Это позволяет всегда присоединять выбранный узел через `tail.next` и не выделять первый элемент результата отдельно. Фиктивный узел в итоговый ответ не входит.

В варианте без фиктивного узла сначала выбираем меньший из первых узлов: он становится головой результата. Затем так же присоединяем остальные узлы.

Оба алгоритма переиспользуют существующие узлы исходных списков, а не создают новые узлы с теми же значениями.

### Реализация

```python
class ListNode:
    def __init__(self, value, next_node=None):
        self.value = value
        self.next = next_node


def merge_with_dummy(list1, list2):
    dummy = ListNode(0)
    tail = dummy

    while list1 is not None and list2 is not None:
        if list1.value <= list2.value:
            tail.next = list1
            list1 = list1.next
        else:
            tail.next = list2
            list2 = list2.next

        tail = tail.next

    if list1 is not None:
        tail.next = list1
    else:
        tail.next = list2

    return dummy.next


def merge_without_dummy(list1, list2):
    if list1 is None:
        return list2

    if list2 is None:
        return list1

    if list1.value <= list2.value:
        head = list1
        list1 = list1.next
    else:
        head = list2
        list2 = list2.next

    tail = head

    while list1 is not None and list2 is not None:
        if list1.value <= list2.value:
            tail.next = list1
            list1 = list1.next
        else:
            tail.next = list2
            list2 = list2.next

        tail = tail.next

    if list1 is not None:
        tail.next = list1
    else:
        tail.next = list2

    return head


def list_from_values(values):
    head = None
    tail = None

    for value in values:
        new_node = ListNode(value)

        if head is None:
            head = new_node
            tail = new_node
        else:
            tail.next = new_node
            tail = new_node

    return head


def values_from_list(head):
    values = []

    while head is not None:
        values.append(head.value)
        head = head.next

    return values
```

### Оценка сложности

Пусть в первом списке `n` узлов, а во втором — `m` узлов.

- **Время:** `O(n + m)`. Каждый узел рассматривается и присоединяется к результату не более одного раза.
- **Дополнительная память:** `O(1)`. В обоих вариантах используются только несколько указателей. Фиктивный узел — одна постоянная дополнительная переменная.

### Визуализация работы алгоритма

Для `list1 = 1 → 2 → 4` и `list2 = 1 → 3 → 4` (в варианте с фиктивным узлом):

| Шаг | Сравниваемые значения | Узел, добавленный в результат | Результат |
|---:|:---|:---:|:---|
| 1 | 1 и 1 | 1 из `list1` | `1` |
| 2 | 2 и 1 | 1 из `list2` | `1 → 1` |
| 3 | 2 и 3 | 2 из `list1` | `1 → 1 → 2` |
| 4 | 4 и 3 | 3 из `list2` | `1 → 1 → 2 → 3` |
| 5 | 4 и 4 | 4 из `list1` | `1 → 1 → 2 → 3 → 4` |
| 6 | В `list1` узлов нет | Оставшийся 4 из `list2` | `1 → 1 → 2 → 3 → 4 → 4` |

### Тесты

Каждый тест запускается для обоих вариантов. Проверим пример из условия, пустые списки, один пустой список, разные длины и повторяющиеся значения. Для каждого запуска создаём новые списки, потому что алгоритм изменяет ссылки между их узлами.

```python
def check_merge(merge_function, first_values, second_values, expected):
    list1 = list_from_values(first_values)
    list2 = list_from_values(second_values)
    merged_head = merge_function(list1, list2)
    assert values_from_list(merged_head) == expected


for merge_function in (merge_with_dummy, merge_without_dummy):
    check_merge(merge_function, [1, 2, 4], [1, 3, 4], [1, 1, 2, 3, 4, 4])
    check_merge(merge_function, [], [], [])
    check_merge(merge_function, [], [1, 2], [1, 2])
    check_merge(merge_function, [1, 2], [], [1, 2])
    check_merge(merge_function, [1, 5, 9], [2, 3, 10], [1, 2, 3, 5, 9, 10])
    check_merge(merge_function, [1, 1, 2], [1, 1, 3], [1, 1, 1, 1, 2, 3])

print("Все тесты пройдены")
```
